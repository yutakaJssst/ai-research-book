```python
import sys
from pathlib import Path
import pytest

print("Python:", sys.version.split()[0])
print("pytest:", pytest.__version__)
demo_dir = Path("average-demo")
demo_dir.mkdir(exist_ok=True)
```

## 関数とテストをファイルにする

配布している `stats.py` と `test_stats.py` と同じコードです。
次の2つのセルで、`average-demo` フォルダ内にファイルを作ります。


```python
%%writefile average-demo/stats.py
# 第6章6.5節で説明した平均値の関数（コードはWebサポートに掲載）
def calculate_average(numbers):
    """Return the average of a list of numbers."""
    if not numbers:
        raise ValueError("numbers must not be empty")
    return sum(numbers) / len(numbers)
```

```python
%%writefile average-demo/test_stats.py
# 第6章6.5節で説明したAI生成テスト（7件のうち、正常、境界、異常を代表する3件）
import pytest
from stats import calculate_average


def test_average_normal():
    # normal case: average of three values
    assert calculate_average([1.0, 2.0, 3.0]) == 2.0


def test_average_single():
    # boundary case: a single value
    assert calculate_average([5.0]) == 5.0


def test_average_empty():
    # error case: empty list raises ValueError
    with pytest.raises(ValueError):
        calculate_average([])
```

## 3件のテストを実行する

`3 passed` と表示されることを確認してください。
正常な入力の平均値、値が1つのとき、空のリストのときを試しています。


```python
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "pytest", "-q", "test_stats.py"],
    cwd=demo_dir,
    check=True,
    capture_output=True,
    text=True,
)
print(result.stdout)
```
